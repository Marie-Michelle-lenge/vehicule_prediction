```python
import pandas as pd
import numpy as np

df=pd.read_csv("C:/Users/LE BON MEC/Documents/cours MM/TFC/data/dataset_DGI_clean_final.csv")

df.head()
```

```python
current_year = 2026

df['age_vehicule'] = current_year - df['annee_fabrication']
```

```python
df['charge_fiscale'] = (
    df['taxe_circulation_usd'] +
    df['taxe_stationnement_usd']
)
```

```python
df['score'] = df['charge_fiscale'] / df['valeur_estimee_usd']
```

```python
df['score'].describe()
```

```python
df['score_final'] = df['score'] * df['age_vehicule']
```

```python
df = pd.get_dummies(df, columns=[
    'type_vehicule',
    'usage',
    'marque'
], drop_first=True)
```

```python
df.to_csv("C:/Users/LE BON MEC/Documents/cours MM/TFC/data/dataset_DGI_featured.csv", index=False)
```
