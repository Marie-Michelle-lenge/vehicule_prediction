```python
df.duplicated().sum()
df = df.drop_duplicates()
#pour le nombre de valeurs dupliquées et pour les supprimer
```

```python
df = df.drop_duplicates()
#supprimer la colonne plaque qui n'a pas d'importance
```

```python
df["nif"] = df["nif"].astype("Int64").astype(str)
#changer le type de données de la colonne nif en entier puis en chaîne de caractères pour éviter les problèmes de formatage
```

```python
df["annee_fabrication"] = df["annee_fabrication"].astype("Int64")
#changer le type de données de la colonne annee_fabrication en entier pour éviter les problèmes de formatage
```

```python
df["usage"] = df["usage"].fillna("Non spécifié")
df["assurance"] = df["assurance"].fillna("Inconnu")
# supprimer lignes sans plaque
df = df.dropna(subset=["plaque_immatriculation"])
# supprimer lignes sans date
df = df.dropna(subset=["date_enregistrement"])
df["controle_technique"] = df["controle_technique"].fillna("Inconnu")
df["vignette"] = df["vignette"].fillna("Inconnu")
#remplacer les valeurs manquantes dans les colonnes usage, assurance, controle_technique et vignette par des valeurs par défaut
```

```python
# remplacer année fabrication par la médiane
df["annee_fabrication"] = df["annee_fabrication"].fillna(df["annee_fabrication"].median())
# remplacer valeur estimée par la médian
df["valeur_estimee_usd"] = df["valeur_estimee_usd"].fillna(df["valeur_estimee_usd"].median())
# remplacer taxe stationnement par 0
df["taxe_stationnement_usd"] = df["taxe_stationnement_usd"].fillna(0)
```

```python
# Numérique
df["puissance"].fillna(df["puissance"].median(), inplace=True)
df["valeur_estimee_usd"].fillna(df["valeur_estimee_usd"].median(), inplace=True)

# Catégoriel
for col in ["type_vehicule", "usage", "marque", "modele", "genre", "couleur"]:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df["valeur_estimee_usd"] = df["valeur_estimee_usd"].astype(str).str.replace(",", ".").astype(float)
# remplacer les virgules par des points et convertir en float pour la colonne valeur_estimee_usd
df["taxe_stationnement_usd"] = df["taxe_stationnement_usd"].astype(str).str.replace(",", ".").astype(float)
# remplacer les virgules par des points et convertir en float pour la colonne taxe_stationnement_usd
df["annee_fabrication"] = df["annee_fabrication"].astype(int)
# convertir en entier pour la colonne annee_fabrication
df["taxe_circulation_usd"] = df["taxe_circulation_usd"].astype(str).str.replace(",", ".").astype(float)
# remplacer les virgules par des points et convertir en float pour la colonne taxe_circulation_usd
```

```python
df = df[df["annee_fabrication"] > 1980]
df = df[df["puissance"] > 0]
df = df[df["valeur_estimee_usd"] > 100]
```
