```python
import pandas as pd

df = pd.read_csv("C:/Users/LE BON MEC/Documents/cours MM/TFC/data/dataset_DGI_featured.csv")

df.head()
```

```python
df['score_final'].describe()
```

```python
q25 = df['score_final'].quantile(0.25)
q55 = df['score_final'].quantile(0.55)
```

```python
def classify(score):
    if score <= q25:
        return "Conforme"
    elif score <= q55:
        return "Non_conforme"
    else:
        return "A_risque"

df['statut_fiscal'] = df['score_final'].apply(classify)
```

```python
df['statut_fiscal'].value_counts(normalize=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['statut_fiscal_encoded'] = le.fit_transform(df['statut_fiscal'])
```

```python
df.to_csv("C:/Users/LE BON MEC/Documents/cours MM/TFC/data/dataset_DGI_ready_ml.csv", index=False)
```
